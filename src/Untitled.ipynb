```python
import os
from ChefsHatGym.gameRooms.chefs_hat_room_local import ChefsHatRoomLocal
from ChefsHatGym.env import ChefsHatEnv
from ChefsHatGym.agents.agent_random import AgentRandon

def main():
    room = ChefsHatRoomLocal(
        "debug_room",
        game_type=ChefsHatEnv.GAMETYPE["MATCHES"],
        stop_criteria=1,            # 1 match so logs are readable
        max_rounds=-1,
        verbose_console=True,       # show in terminal
        verbose_log=True,           # write to file
        game_verbose_console=True,  # detailed game logs in terminal
        game_verbose_log=True,      # detailed game logs in file
        save_dataset=False,
    )

    log_dir = room.get_log_directory()
    print("Room log directory:", log_dir)

    # IMPORTANT: set players verbose_log=True too
    for i in range(4):
        room.add_player(AgentRandon(f"R{i}", log_dir, verbose_log=True))

    summary = room.start_new_game()
    print("Finished. Summary keys:", list(summary.keys()) if isinstance(summary, dict) else type(summary))

    # Try to locate the log file in temp folder
    # Usually something like temp\xx-xxxx_debug_room\Log\Log.log
    print("\nLook for log files under this project folder:")
    print(os.path.abspath("temp"))

if __name__ == "__main__":
    main()
```
