```python
from ChefsHatGym.gameRooms.chefs_hat_room_local import ChefsHatRoomLocal
from ChefsHatGym.env import ChefsHatEnv
from ChefsHatGym.agents.agent_random import AgentRandon

room = ChefsHatRoomLocal(
    "test_room",
    game_type=ChefsHatEnv.GAMETYPE["MATCHES"],
    stop_criteria=1,
    max_rounds=-1,
    verbose_console=False,
    verbose_log=False,
    game_verbose_console=False,
    game_verbose_log=False,
    save_dataset=False,
)

log_dir = room.get_log_directory()
for i in range(4):
    room.add_player(AgentRandon(f"A{i}", log_dir, verbose_log=False))

summary = room.start_new_game()
print("Score:", summary.get("score"))
print("Performance Score:", summary.get("performanceScore"))
```

```python
print("Summary type:", type(summary))
print("Summary:", summary)
if isinstance(summary, dict):
    print("Keys:", list(summary.keys()))
```

```python
def get_metrics(summary):
    if not isinstance(summary, dict):
        return {"score": None, "performanceScore": None}

    # common possibilities across versions
    score = (
        summary.get("score")
        or summary.get("Score")
        or summary.get("scores")
        or summary.get("Scores")
        or summary.get("matchScore")
        or summary.get("Match_Score")
    )

    perf = (
        summary.get("performanceScore")
        or summary.get("Performance_Score")
        or summary.get("performance_score")
        or summary.get("PerformanceScore")
        or summary.get("gamePerformanceScore")
        or summary.get("Game_Performance_Score")
    )
    return {"score": score, "performanceScore": perf}

metrics = get_metrics(summary)
print(metrics)
```
